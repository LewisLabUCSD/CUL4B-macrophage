```python
# update SIG xlsx to include shared dysregulated proteins/phosphoproteins in at least 2-3 timepoints
import pandas as pd
import numpy as np
```

```python
anova = {}
anova['PROT'] = pd.read_excel('../data/PROT_ANOVA.xlsx',index_col=0)
anova['PHOS'] = pd.read_excel('../data/PHOS_ANOVA.xlsx',index_col=0)

# prot_sig = pd.read_excel('../data/PROT_SIG.xlsx',sheet_name=None)
# phos_sig = pd.read_excel('../data/PHOS_SIG.xlsx',sheet_name=None)
```

```python
shared_2 = {}
shared_3 = {}
    
for p in [*anova]:
    pdf = anova[p].copy()
    shared_2[p] = pdf[((pdf[tp].abs() >= 0.58).sum(axis=1) >=2) & (pdf['ADJPVAL'] < 0.05)]
    shared_2[p]['sig_timepoints'] = shared_2[p].apply(lambda row: [i.split('_')[1] for i in [col for col in tp if np.abs(row[col]) >= 0.58]], axis=1)

    shared_3[p] = pdf[((pdf[tp].abs() >= 0.58).sum(axis=1) >=3) & (pdf['ADJPVAL'] < 0.05)]
    shared_3[p]['sig_timepoints'] = shared_3[p].apply(lambda row: [i.split('_')[1] for i in [col for col in tp if np.abs(row[col]) >= 0.58]], axis=1)
```

```python
shared_3[p]
```

```python
sig_anova = {}
for p in [*anova]:
    sig_anova[p] = anova[p][anova[p]['ADJPVAL'] < 0.05]
```

```python
for p in [*anova]:
    file_path='../data/'+p+'_SIG.xlsx'
    
    # Append DataFrame to a new sheet in the existing file
    with pd.ExcelWriter(file_path, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
        shared_2[p].to_excel(writer, sheet_name='shared_2', index=True)
        shared_3[p].to_excel(writer, sheet_name='shared_3', index=True)
        sig_anova[p].to_excel(writer, sheet_name='SIG_DYSREG', index=True)
```
